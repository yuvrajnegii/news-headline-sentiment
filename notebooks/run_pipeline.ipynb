```python
import subprocess
import sys
from pathlib import Path

# Project root, whether this notebook sits in notebooks/ or in the root folder
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

def run(script):
    print(f"=== running {script} ===")
    result = subprocess.run([sys.executable, str(ROOT / "src" / script)],
                            cwd=ROOT, capture_output=True, text=True)
    print(result.stdout)
    if result.returncode != 0:
        print(result.stderr)
        raise RuntimeError(f"{script} failed (see the error above)")
    print(f"=== {script} finished OK ===")

for f in ["data/raw/videos.csv", "data/raw/sentiments.csv"]:
    print(("OK       " if (ROOT / f).exists() else "MISSING  ") + f)
```
